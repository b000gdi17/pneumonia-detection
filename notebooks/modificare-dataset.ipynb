```python
import os
import shutil
import numpy as np
from sklearn.model_selection import train_test_split

base_dir = '../dataset/chest_xray/train'         
output_base = '../dataset/chest_xray_split'       

images, labels = [], []
for category in ['NORMAL', 'PNEUMONIA']:
    category_dir = os.path.join(base_dir, category)
    for img_name in os.listdir(category_dir):
        img_path = os.path.join(category_dir, img_name)
        if os.path.isfile(img_path) and img_name.lower().endswith(('.jpeg', '.jpg', '.png')):
            images.append(img_path)
            labels.append(category)


train_imgs, val_imgs, train_lbls, val_lbls = train_test_split(
    images, labels,
    test_size=0.20,
    stratify=labels,
    random_state=42
)

for split in ['train', 'val']:
    for category in ['NORMAL', 'PNEUMONIA']:
        os.makedirs(os.path.join(output_base, split, category), exist_ok=True)

for img_path, label in zip(train_imgs, train_lbls):
    dest = os.path.join(output_base, 'train', label, os.path.basename(img_path))
    shutil.copy(img_path, dest)

for img_path, label in zip(val_imgs, val_lbls):
    dest = os.path.join(output_base, 'val', label, os.path.basename(img_path))
    shutil.copy(img_path, dest)

print(f"Train: {len(train_imgs)} images")
print(f"Val:   {len(val_imgs)} images")
print("Split realizat! Verifică folderul: chest_xray_split/")
```
